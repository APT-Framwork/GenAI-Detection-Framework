```python
import pandas as pd

# File paths
input_path = r'20-October-P2.csv'
output_path = r'20-October-P2_filtered.csv'
dropped_path = r'20-October-P2_dropped.csv'

print("="*100)
print("FILTERING CSV FILE - KEEPING ONLY ROWS STARTING WITH 'wazuh-alerts-'")
print("="*100)

# Step 1: Read the CSV file
print("\n📂 Step 1: Reading CSV file...")
print("-"*100)

# Read the CSV file (skip bad lines if any)
df = pd.read_csv(input_path, on_bad_lines='skip', encoding='utf-8', engine='python')

print(f"✅ File loaded: {input_path}")
print(f"   Total rows: {len(df):,}")
print(f"   Total columns: {len(df.columns)}")

# Display first column name
first_col = df.columns[0]
print(f"\n📌 First column name: {first_col}")

# Step 2: Analyze the data
print(f"\n{'='*100}")
print("🔍 Step 2: Analyzing first column values...")
print("-"*100)

# Show first 10 values in the first column
print(f"\n🔸 First 10 values in column '{first_col}':")
for i, val in enumerate(df[first_col].head(10), 1):
    print(f"   {i}. {val}")

# Step 3: Filter rows that start with 'wazuh-alerts-'
print(f"\n{'='*100}")
print("🔧 Step 3: Filtering rows...")
print("-"*100)

# Create mask for rows starting with 'wazuh-alerts-'
mask_keep = df[first_col].astype(str).str.startswith('wazuh-alerts-')

# Separate dataframes
df_keep = df[mask_keep].copy()
df_drop = df[~mask_keep].copy()

print(f"\n📊 Results:")
print(f"   ✅ Rows starting with 'wazuh-alerts-': {len(df_keep):,} ({len(df_keep)/len(df)*100:.2f}%)")
print(f"   ❌ Rows NOT starting with 'wazuh-alerts-': {len(df_drop):,} ({len(df_drop)/len(df)*100:.2f}%)")
print(f"   📊 Total rows: {len(df):,}")

# Step 4: Show samples
print(f"\n{'='*100}")
print("📋 Step 4: Sample of filtered data:")
print("-"*100)

print(f"\n✅ Sample of KEPT rows (starting with 'wazuh-alerts-'):")
if len(df_keep) > 0:
    print(df_keep[[first_col]].head(5))
else:
    print("   No rows found starting with 'wazuh-alerts-'")

print(f"\n❌ Sample of DROPPED rows (NOT starting with 'wazuh-alerts-'):")
if len(df_drop) > 0:
    print(df_drop[[first_col]].head(5))
else:
    print("   No rows to drop!")

# Step 5: Save the filtered data
print(f"\n{'='*100}")
print("💾 Step 5: Saving filtered data...")
print("-"*100)

# Save the kept rows (starting with 'wazuh-alerts-')
df_keep.to_csv(output_path, index=False)
print(f"✅ Filtered file saved: {output_path}")
print(f"   Rows: {len(df_keep):,}")

# Save the dropped rows (optional - for verification)
if len(df_drop) > 0:
    df_drop.to_csv(dropped_path, index=False)
    print(f"\n📋 Dropped rows saved: {dropped_path}")
    print(f"   Rows: {len(df_drop):,}")
    print(f"   (This file is for reference - shows what was filtered out)")
else:
    print(f"\n✅ No rows to drop - all rows start with 'wazuh-alerts-'!")

# Step 6: Verification
print(f"\n{'='*100}")
print("✔️ Step 6: Verification:")
print("-"*100)

# Verify the new file
df_verify = pd.read_csv(output_path, encoding='utf-8', engine='python')
print(f"✅ Verification of filtered file:")
print(f"   Rows: {len(df_verify):,}")
print(f"   Columns: {len(df_verify.columns)}")

# Check if all rows start with 'wazuh-alerts-'
all_start_with = df_verify[first_col].astype(str).str.startswith('wazuh-alerts-').all()
print(f"\n   All rows start with 'wazuh-alerts-': {'✅ YES' if all_start_with else '❌ NO'}")

if all_start_with:
    print(f"   ✅ Perfect! All rows in the new file start with 'wazuh-alerts-'")
else:
    not_starting = (~df_verify[first_col].astype(str).str.startswith('wazuh-alerts-')).sum()
    print(f"   ⚠️ Warning: {not_starting} row(s) don't start with 'wazuh-alerts-'")

# Show first 5 rows of verified file
print(f"\n🔸 First 5 rows of filtered file:")
print(df_verify[[first_col]].head())

print(f"\n{'='*100}")
print("✨ PROCESS COMPLETE!")
print("="*100)

print(f"\n📊 FINAL SUMMARY:")
print(f"   Original file: {input_path}")
print(f"   Original rows: {len(df):,}")
print(f"\n   Filtered file: {output_path}")
print(f"   Filtered rows: {len(df_keep):,} (kept rows starting with 'wazuh-alerts-')")
print(f"   Rows removed: {len(df_drop):,}")
print(f"\n   Dropped rows file: {dropped_path}")
print(f"   (Contains rows that don't start with 'wazuh-alerts-' - for reference)")

print(f"\n{'='*100}")
print("🎉 SUCCESS! File filtered successfully!")
print(f"   ✅ {len(df_keep):,} rows starting with 'wazuh-alerts-' saved to: {output_path}")
print("="*100)
```
